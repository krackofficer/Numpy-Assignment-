```python
## Q1. Create a 3x3 NumPy array with random integers between 1 and 100. Then, interchange its rows and columns.
```

```python

import numpy as np

# Create a 3x3 array with random integers
array = np.random.randint(1, 101, size=(3, 3))
print("Original Array:")
print(array)

# Interchange rows and columns
transposed_array = array.T
print("Transposed Array:")
print(transposed_array)
```

```python
# 2. Generate a 1D NumPy array with 10 elements. Reshape it into a 2x5 array, then into a 5x2 array.
```

```python
array_2 = np.arange(10)
print("\nOriginal 1D Array:")
print(array_2)

array_2_2x5 = array_2.reshape(2,5)
print("Reshaped to 2x5:")
print(array_2_2x5)

array_2_5x2 = array_2.reshape(5,2)
print("Reshaped to 5x2:")
print(array_2_5x2)
```

```python
# 3. Create a 4x4 NumPy array with random float values. Add a border of zeros around it, resulting in a 6x6 array.
```

```python
array_3 = np.random.rand(4,4)
print("\nOriginal 4x4 Array:")
print(array_3)

bordered_array_3 = np.pad(array_3, pad_width=1, mode='constant', constant_values=0)
print("6x6 Array with Border:")
print(bordered_array_3)
```

```python
# 5. Create a NumPy array of strings ['python', 'numpy', 'pandas']. Apply different case transformations.
```

```python
array_5 = np.array(['python', 'numpy', 'pandas'])

uppercase_5 = np.char.upper(array_5)
lowercase_5 = np.char.lower(array_5)
titlecase_5 = np.char.title(array_5)

print("\nUppercase :", uppercase_5)
print("Lowercase :", lowercase_5)
print("Titlecase :", titlecase_5)
```

```python
# 9. Generate a NumPy array of 100 random integers between 0 and 1000. Find and display all prime numbers in this array.
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(np.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

array_9 = np.random.randint(0, 1001, size=100)
prime_numbers_9 = np.array([num for num in array_9 if is_prime(num)])

print("\nOriginal Array :")
print(array_9)
print("Prime Numbers :")
print(prime_numbers_9)
```

```python
# 10. Create a NumPy array representing daily temperatures for a month. Calculate and display the weekly averages.
```

```python
temperatures_10 = np.random.uniform(low=20.0, high=40.0, size=30)
weekly_avg_10 = temperatures_10.reshape(5,6).mean(axis=1)

print("\nDaily Temperatures for a Month :")
print(temperatures_10)
print("Weekly Averages :")
print(weekly_avg_10)
```
